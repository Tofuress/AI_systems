```python
import yfinance as yf
import matplotlib.pyplot as plt
```

```python
tickers = ['AAPL', 'MSFT', 'GOOG']
data = yf.download(tickers, start='2021-01-01', end='2022-01-01')['Close']
data
```

```python
data_norm = data / data.iloc[0] * 100

fig, ax = plt.subplots(figsize=(12, 6))
ax.plot(data_norm['AAPL'], color='red', label='AAPL')
ax.plot(data_norm['MSFT'], color='blue', label='MSFT')
ax.plot(data_norm['GOOG'], color='green', label='GOOG')
ax.set_xlabel('Дата')
ax.set_ylabel('Относительный рост')
plt.show()
```
